```python
# Pluralization 
pairs = [
    ("cat", "cats"),
    ("dog", "dogs"),    
    ("car", "cars"),
    ("bus", "buses"),
    ("baby", "babies"),
]

dirs = [vec(p2) - vec(p1) for p1, p2 in pairs]
dirs = torch.stack(dirs)
plural = dirs.mean(dim = 0) # show first 10 dims of the average plural vector
```

```python
for i, (p1, p2) in enumerate(pairs): 
    print(p1, "->", p2, cos(dirs[i], plural).item())
```

```python
# Clustering in the latent space 
from sklearn.decomposition import PCA 
import matplotlib.pyplot as plt
import itertools as it 

words = ["dog", "cat", "car", "truck", "bus", "bicycle"]

X = torch.stack([vec(w) for w in words]).numpy()

p = PCA(n_components=2).fit_transform(X)
```

```python
plt.scatter(*p.T)
for (x, y), w in zip(p, words): 
    plt.text(x, y, w)

for (x1, y1), (x2, y2) in zip(p[::2], p[1::2]): 
    plt.plot([x1, x2], [y1, y2], "k--", alpha=0.5`)
```
