```python
from scipy.stats import describe, norm, linregress; import numpy as np
# help(norm.rvs)

#SciPy Stats 
print("---------SciPy Stats Module Example---------\n")

#generate random data from a stanrdard normal distribution
data = norm.rvs(loc=0, scale=1, size=1000)
#fit distribution to data 
mean, variance = norm.fit(data)
print(f"Mean: {mean}, Variance: {variance}")

#SciPy Descriptive Statistics 
print("\n---------SciPy Descriptive Statistics Example---------\n")
desc = describe(data); print(type(desc))
print("Descriptive Statistics:")
print(f"nobs: {desc.nobs}, minmax: {desc.minmax}, mean: {desc.mean}, variance: {desc.variance}, skewness: {desc.skewness}, kurtosis: {desc.kurtosis}")

#SciPy Statistics Models 
print("\n---------SciPy Statistics Models Example---------\n")
math_score = np.array([88, 92, 80, 89, 100, 94, 91, 85, 87, 95])
phys_score = np.array([84, 90, 78, 88, 96, 92, 89, 82, 85, 91])
slope, intercept, r_value, p_value, std_err = linregress(math_score, phys_score)
print("\nLinear Regression Results:")
print(f"Slope: {slope}, Intercept: {intercept}, R-squared: {r_value**2}, P-value: {p_value}, Standard Error: {std_err}")
```
