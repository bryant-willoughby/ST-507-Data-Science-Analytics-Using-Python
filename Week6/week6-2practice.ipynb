### Problem 1 (revisit)
Use numpy to create a function that adds a border filled with zeros around an existing array.

*Hint*: Use numpy's `pad` function. See https://numpy.org/doc/stable/reference/generated/numpy.pad.html

Note: we did not cover this function in class, see if you are comfortable of reading numpy documents and learn on your own!

```python
def add_border(x):
    return np.pad(x, 1, 'constant', constant_values=0)
```

```python
x = np.ones((1, 1))
claimed = add_border(x)
correct = np.zeros((3, 3))
correct[1, 1] = 1.0
assert claimed.shape == correct.shape
assert np.all(np.abs(claimed - correct) < 1e-8)
```

```python
# Let's go throught the document together
# The np.pad function in NumPy is versatile and can be used in various ways to add borders or padding to arrays
# 1) Number of values padded to the edges of each axis. ((before_1, after_1), ... (before_N, after_N)) 
x = np.array([[1, 2], [3, 4]])
padded = np.pad(x, pad_width=((1,2), (2,1)), mode='constant', constant_values=0)
print(padded)
```

```python
# Pad with different mode 'constant' -> 'mode'
x = np.array([[1, 2], [3, 4]])
padded = np.pad(x, pad_width=((1,2), (2,1)), mode='edge')
print(padded)
```

```python
# Pad with different values on each side
x = np.array([[1, 2], [3, 4]])
padded = np.pad(x, pad_width=((1,2), (2,1)), mode='constant', constant_values=((5,6), (7,8)))
print(padded)
```

```python
def create_city_temperatures(temperatures, cities, offsets=None):
    """
    Create a 2D array (n_cities x n_days) of temperatures for multiple cities.
    All cities share the same daily pattern (temperatures) but have different baselines (offsets).

    Default offsets for 3 cities: [0, +5, -2].
    """
    if not offsets: 
        offsets = np.array([0, 5, -2])
    temperatures = temperatures.reshape(1,len(temperatures))
    offsets = offsets.reshape(len(cities),1)
    return temperatures + offsets

# def create_city_temperatures(temperatures, cities, offsets=None):
#     """
#     Create a 2D array (n_cities x n_days) of temperatures for multiple cities.
#     All cities share the same daily pattern (temperatures) but have different baselines (offsets).

#     Default offsets for 3 cities: [0, +5, -2].
#     """
#     temps = np.asarray(temperatures, dtype=float)  # ensure float for offsets
#     cities = np.asarray(cities)
#     n_cities = cities.shape[0]

#     if offsets is None:
#         # Default per prompt: City1 +0, City2 +5, City3 -2; extra cities get 0
#         offsets = np.zeros(n_cities, dtype=float)
#         if n_cities >= 2:
#             offsets[1] = 5.0
#         if n_cities >= 3:
#             offsets[2] = -2.0
#     else:
#         offsets = np.asarray(offsets, dtype=float)
#         if offsets.shape[0] != n_cities:
#             raise ValueError("offsets length must match number of cities")

#     # Broadcast: (n_cities, 1) + (1, n_days) -> (n_cities, n_days)
#     return offsets[:, np.newaxis] + temps[np.newaxis, :]
```

```python
## More on array manipulation 

# More on broadcasting 
    #np.newaxis and None are used to add a new dimension to an existing array 

#expanding vector dimensions using np.newaxis for compatibility
print("\nExpanding dimensions with np.newaxis:")
my_matrix = np.array([[1, 2, 3],
                       [4, 5, 6]]); print(my_matrix); print(my_matrix.shape) # (2,3)
my_vec = np.array([10, 20]); print(my_vec); print(my_vec.shape) # (2,)
my_vec = my_vec[:, np.newaxis]; print(my_vec); print(my_vec.shape) # (2,1)

print(my_matrix + my_vec) # (2,3) + (2,1) -> (2,3)

#expanding vector dimensions using None 
print("\nExpanding dimensions with None:")
my_matrix = np.array([[1, 2, 3],
                       [4, 5, 6]]); print(my_matrix); print(my_matrix.shape) # (2,3)
my_vec = np.array([10, 20]); print(my_vec); print(my_vec.shape) # (2,)
my_vec = my_vec[:, None]; print(my_vec); print(my_vec.shape) # (2,1)

print(my_matrix + my_vec) # (2,3) + (2,1) -> (2,3)
```

```python
# More on matrix multiplication 
    # @ operator preferred 

#create two 2D NumPy arrays 
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

#matrix multiplication using np.matmul()
print("Matrix multiplication using np.matmul():")
print(np.matmul(A, B))

#matrix multiplication using @ operator
print("Matrix multiplication using @ operator:")
print(A @ B)
```

```python
# Array Stack 
a = np.array([[1,2],[3,4]], dtype = 'uint'); print(a)
b = np.array([[5,6],[7,8]], dtype = 'uint'); print(b)
print("----------- Vertical Stack -----------\n")
print(np.vstack((a, b)))
print("----------- Horizontal Stack -----------\n")
print(np.hstack((a, b)))
print("----------- Concatenate Vertical -----------\n")
print(np.concatenate((a,b), axis=0)) # vertical stack
print("----------- Concatenate Horizontal -----------\n")
print(np.concatenate((a,b), axis=1)) # horizontal stack
```
