#### 3.1 Dict

```python
def get_student_grade(student_grades, student_name):
    grade = student_grades[student_name] #KeyError if student_name not in dictionary
    return f"{student_name}'s grade is {grade}"

grades = {"Alice": 95, "Bob": 87}
print(get_student_grade(grades, "Alice"))
print(get_student_grade(grades, "Diana"))
```

```python
def get_student_grade(student_grades, student_name):
    '''
    Get student grade by name
    Input: student_grades (dict): Dictionary of student names to grades
         Key: student name (str)
         Value: grade (int)
    '''
    if student_name in student_grades: # check if student_name exists in dictionary
        grade = student_grades[student_name]
        return f"{student_name}'s grade is {grade}"
    else: 
        return f"{student_name} is not in the gradebook."

grades = {"Alice": 95, "Bob": 87}
print(get_student_grade(grades, "Alice"))
print(get_student_grade(grades, "Diana"))
```

#### 3.2 Handle Corner case.

```python
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers) # check for division by zero
    average = total / count
    return average

print(calculate_average([10, 20, 30]))
print(calculate_average([]))
```

```python
def calculate_average(numbers):
    #handling this zero division corner case 
    if len(numbers) == 0:
        return 0  # or return None, or raise an exception
    else: 
        total = sum(numbers)
        count = len(numbers) # check for division by zero
        assert count != 0, "Count should not be zero here"
        average = total / count
        return average

print(calculate_average([10, 20, 30]))
print(calculate_average([]))
```
