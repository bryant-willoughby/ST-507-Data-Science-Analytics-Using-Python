### Problem 1: Filtering and computing a mean
Write a function that returns the average rainfall on days with wind speed above 6 mph.

```python
# ff.columns
# ff[ff['wind'] > 6]
# ff['rain'].mean()

def windy_rain():
    '''Returns average rainfall on days with wind speed above 6 mph'''
    windy_days = ff[ff['wind'] > 6] #filter for windy days
    return windy_days['rain'].mean() #average rainfall on windy days
    
```

```python
assert abs(windy_rain() - 0.0412698) < 1e-6
```

### Problem 2: Filtering and computing a sum
Write a function that take a month as an argument and returns the total rainfall during the month.

```python
# ff['month'].unique()
# ff['month'] == 'mar'
# ff[ff['month'] == 'mar'].shape
# ff['rain'].sum()

def monthly_rain(month):
    # filter data for the specified month
    month_data = ff[ff['month'] == month]

    #calculate total rainfall for the month
    return month_data['rain'].sum()
```

```python
assert abs(monthly_rain("aug") - 10.8) < 1e-6
assert abs(monthly_rain("mar") - 0.2) < 1e-6
```

Download the white wine dataset from [https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv](https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv), and load it into a Pandas dataframe called `df`.

### Problem 4: pipe
Write a function that takes a user-defined function as an argument.
It uses `pipe` to apply the function to each wine pH, and returns the mean of the values returned by the user-defined function.

```python
# df['pH']

def mean_ph(user_fun):
    '''Takes user-defined function; uses pipe to apply function to each wine pH and returns the mean'''
    return df['pH'].pipe(user_fun).mean()
```

```python
assert abs(mean_ph(lambda x: x**2) - (df.pH ** 2).mean()) < 1e-8
```

### Pandas Series Basics

```python
import pandas as pd 
import numpy as np 

# Series Basics 
print("------ Series Basics ------\n")
numbers = np.random.randn(5)
s = pd.Series(numbers)
print(s)

# Series Creation 
print("\n------ Series Creation ------\n")
idx = ['a', 'b', 'c', 'd', 'e']
s2 = pd.Series(numbers, index=idx)
print(s2)

print("\n------Series Creation (from dictionary) ------\n")
d = {'dog':3.1415, 'cat':42, 'bird':0, 'goat':1.618, 'cthulu':np.nan}
s = pd.Series(d); print(s)

print("\n------ Series Indexing ------\n")
inds = ['dog', 'cat', 'bird', 'goat', 'chutlu'] # chutlu not in d; NaN result
s = pd.Series(d, index=inds); print(s)
```

```python
print("\n------ Series properties ------\n")
numbers = np.array([2, 50.0, 113.0, 4, 9,])
s = pd.Series(numbers, name = 'sale'); print(s)
print("\n Common attributes:")
print(f"Name: {s.name}")
print(f"dtype: {s.dtype}")
print(f"Index: {s.index}")
print(f"Size: {s.size}")
print(f"Shape: {s.shape}")
print(f"Values: {s.values}")
```

### Pandas Series Operations

```python
print("\n------ Indexing ------\n")
s = pd.Series([2, 3, 5, 7, 11])
print(s)
print(s[0]) # first element
print(s[1:3]) # slice elements 1 and 2
# s[-1] # doesn't work

print("\n ----- Non-Unique Indices -----\n")
s = pd.Series([1, 2, 3, 2, 1], index=['a','a', 'a', 'a', 'a'])
print(s['a']) #indices need not be unique
```

```python
print("\n------ Mapping and linking Series values ------\n")
s = pd.Series(['dog', 'goat', 'skunk']); print(s)
s.map(lambda s: len(s)) #takes function and applies to each element

s = pd.Series(['fruit', 'animal', 'animal', 'fruit', 'fruit'], 
              index = ['apple', 'cat', 'goat', 'banana', 'kiwi'])
print(s)

t = pd.Series({'fruit': 0, 'animal': 1})
s.map(t)
```

### Pandas Data Frames 

```python
print("------ Pandas Data Frames Creation ------")
d = {'A': pd.Series([1, 2, 3], index=['cat', 'dog', 'bird']), 
     'B':{'cat': 3.14, 'dog': 2.71, 'bird': 1.41, 'goat': 1.73}}
df = pd.DataFrame(d)
print(df)

d = {'Undergrad': pd.Series(['Umich', 'Stanford', 'Princeton', 'Columbia'], 
                            index = ['Ford', 'Hoover', 'Wilson', 'Obama']), 
    'PhD': {'Wilson': 'Johns Hopkins'}, 
    'JD': {'Ford': 'Yale', 'Obama': 'Harvard'}, 
    'Terms': pd.Series([1,1,2,2], index = ['Hoover', 'Ford', 'Wilson', 'Obama'])}
presidents = pd.DataFrame(d); print("\n",presidents)

print("\n----- row/column as attributes -----")
print(presidents.columns) # Pandas Index object
print(presidents.index)

print("\n----- Accessing/adding columns -----")
print(presidents['PhD']) #access column as dictionary
presidents['Years'] = 4*presidents['Terms'] #add new column
presidents['Nobels'] = [0, 0, 1, 1] #add new column
presidents['Field Medals'] = 0 #scalars broadcast across the rows
print(presidents)

print("\n----- Deleting columns -----")
del presidents['Years']
print(presidents)
fields = presidents.pop('Field Medals') #removes and returns column
print(fields)
```

```python
print("\n----- Indexing & Selection -----")
print(presidents.loc['Obama']) #selects row by label 
print(presidents.iloc[2]) #selects row by integer position
print(presidents['JD']) #select column by name 
print(presidents[1:3])
print(presidents[presidents['Terms'] < 2])
```

```python
print("\n------ Arithmetic -------")
df1 = pd.DataFrame(np.random.randn(8, 4), columns = ['A', 'B', 'C', 'D'])
df2 = pd.DataFrame(np.random.randn(5, 3), columns = ['A', 'B', 'C'])
print(df1 + df2) #fills non-alignable entries with NaN

df = pd.DataFrame(np.random.randn(4, 2), columns = ['A', 'B'])
print(df)
print(df - df.iloc[0]) #row-wise subtraction (broadcasting)
print(10*df + 1)
print(df > 0)
```

```python
print("\n----- Arithmetic with DataFrames -----")
print(df)
print((df > 0).any())
print((df > 0).all())
# df or df # unsupported as also seen in NumPy
print(df.values) #get underlying NumPy array
print(df.T) #transpose of DataFrame
print(df.T * df.T) #entrywise multiplication 
print(df.T.dot(df)) #matrix multiplication
```

```python
print("\n------ Removing NaNs -------")
df['C'] = [1, np.nan, 3, np.nan]
df['D'] = [np.nan, np.nan, np.nan, np.nan]
print(df)
print(df.dropna(axis=1, how='any')) #drops columns with any NaNs
print(df.dropna(axis=1, how='all')) #drops columns with all NaNs
```

```python
print("\n------ Comparing DataFrames -------")
print(df)
df1 = 2*df
df2 = df + df
(df1 == df2).all()
print(np.nan == np.nan) #False; NaN not equal to NaN by definition
print(df1.equals(df2)) #True

print("\n----- Statistical Operations on DataFrames -----")
print(df)
print(np.nanmean(np.array(df.iloc[1]))) #ignores NaN; possible but tedious w/ NumPy
print(df.mean(axis = 0)) #column-wise mean; ignores NaN by default
print(df.mean(axis = 1)) #row-wise mean; ignores NaN by default
```

```python
print("\n----- Row and Column-wise function: apply() -----")
print(df.apply(np.mean)) #column-wise mean; default axis=0
print(df.apply(np.mean, axis=1)) #row-wise mean

def quadratic(x, a, b, c=1): 
    return a*x**2 + b*x + c
df.apply(quadratic, args=(2, 3), c=3) #applies quadratic to each column
```
