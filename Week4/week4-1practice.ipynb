### Question 1
Create a Rectangle class with the following specifications:
Represents a rectangle whoses sides are parallel to the x and y axes. Specified by its upper-left corner, height, and width.

Add an area() method to return the area of the rectangle.

```python
class Point():
    ''' Point class represents and finds distance between x,y coordinates. '''
    name = "I am a Point class data attibute"
    def __init__(self, xval, yval):
        self.x = xval
        self.y = yval
        
    def __str__(self):
        return f"Point({self.x}, {self.y})"  
        
    def distance(self, other_point):
        x_diff_square = (self.x - other_point.x) ** 2
        y_diff_square = (self.y - other_point.y) ** 2
        return (x_diff_square + y_diff_square) ** 0.5

p1 = Point(0,0)
print(p1)
```

```python
class Rectangle: 
    ''' Rectangle class represents a rectangle defined by an upper left point, height, and width. '''
    def __init__(self, upper_left, height, width):
        self.upper_left = upper_left # upper_left is a Point object
        self.height = height
        self.width = width

    def __str__(self):
        return f"Rectangle(upper_left={self.upper_left}, height={self.height}, width={self.width})"

    def area(self):
        return self.height * self.width

upper_left_corner = Point(0,0)
my_rectangle = Rectangle(upper_left_corner, 2, 4)
print(my_rectangle)
print(my_rectangle.area())
```

```python
# Pure functions and modifiers 
class Time: 
    '''Represents time on a 24 hour clock. 
    Attributes: int, hours, int mins, int secs '''

    def print_time(self):
        print(f"{self.hours}:{self.mins}:{self.secs}")

    def time_to_int(self): #method but still a pure function; it has no side effects 
        return (self.secs + 60*self.mins + 3600*self.hours)
    
    #pure method (function); does not change attributes of caller
    def increment_pure(self, seconds): 
        '''Returns new Time object representing this time 
        incremented by the given number of seconds.'''
        t = Time()
        t = int_to_time(self.time_to_int() + seconds)
        return t
    
    #modifier method; changes attributes of caller
    def increment_modifier(self, seconds):
        '''Increments this time by the given number of seconds.'''
        (mins, self.secs) = divmod(self.secs + seconds, 60)
        (hrs, self.mins) = divmod(self.mins + mins, 60)
        self.hours = (self.hours + hrs) % 24

t1 = int_to_time(1234)
t1.increment_modifier(1111)
t1.time_to_int() #modifier method does indeed change attributes of caller 
# t1.increment_pure(100,200) #caller is considered an argument (because of self)
```

```python
import random 
# Creating Deck class 
class Deck: 
    '''Represents a deck of cards'''
    def __init__(self): 
        self.cards = list()
        for suit in range(4): 
            for rank in range(1,14): 
                card = Card(suit, rank)
                self.cards.append(card)
    
    def __str__(self): 
        res = list()
        for c in self.cards: 
            res.append(str(c))
        return ('\n'.join(res))

    def pop_card(self): 
        return(self.cards.pop())
    
    def add_card(self, card):
        self.cards.append(card)
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def move_cards(self, hand, ncards): 
        for i in range(ncards): 
            hand.add_card(self.pop_card())

d = Deck()
print(d)
d.shuffle()
print(d)

d = Deck(); d.shuffle()
h = Hand()
d.move_cards(h,5)
print(h)
```
