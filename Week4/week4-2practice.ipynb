### Question 1
Implement a custom range iterator class called MyRange that mimics some of the basic functionality of Python's built-in range() function. This exercise will help you understand how iterators work in Python.

1. Create a class named MyRange that takes two integer arguments in its constructor:
start: the first value of the range
end: the value to stop before (exclusive)

2. The class should implement the iterator protocol, making it usable in a for loop.

3. Each iteration should return the next integer in the sequence, starting from start and incrementing by 1 each time.
The iteration should stop when the value would become equal to or greater than end.

```python
class MyRange: 
    '''Create manual range function that mimics built-in range()'''
    def __init__(self, start, stop=None, step=1):
        # support single-argument usage MyRange(n) -> 0..n-1
        if stop is None:
            stop = start
            start = 0
        if step == 0:
            raise ValueError("step must not be zero")
        # current holds the next value to return
        self.current = start
        # stop is exclusive bound
        self.stop = stop
        # step may be positive or negative
        self.step = step

    def __iter__(self):
        # iterator protocol: an iterator returns itself from __iter__
        return self

    def __next__(self):
        # stop condition depends on sign of step
        if (self.step > 0 and self.current >= self.stop) or (self.step < 0 and self.current <= self.stop):
            raise StopIteration
        value = self.current
        # advance internal state for the next call
        self.current += self.step
        return value
```

```python
# MyRange should work as an iterator
nums = MyRange(1, 5)
for num in nums:
    print(num)
```

```python
nums = MyRange(1, 5)
print(next(nums))
print(next(nums))
print(next(nums))
print(next(nums))
```

### Question 2
Create the same custom iterators in Python using generator-based approach.

```python
def my_range(low,high,step=1): 
    current = low
    while current < high: 
        yield current
        current += step
```

```python
nums = my_range(1, 5)
print(type(nums))
for num in nums:
    print(num)
```

```python
nums = my_range(1, 5)
print(next(nums))
print(next(nums))
print(next(nums))
print(next(nums))
```

```python
# Generators 

#function call 
    #local namespace created; all work dissapears once namespace destroyed
def harmonic(n): 
    return(sum([1/k for k in range(1,n+1)]))
print(harmonic(10))

#generators 
    #maintain internal state; yield instead of return 
def harmonic():
    (h, n) = (0, 1)                # h = running sum, n = next denominator (starts at 1)
    while True:
        (h, n) = (h + 1 / n, n + 1)  # add 1/n to the running sum, then increment n
        yield h                     # yield current partial sum (H1, H2, ...)
h = harmonic()                       # create generator; use next(h) to get successive values
[next(h) for _ in range(10)]
```

```python
# Writing iterators using generators 
def harmonic(): 
    (h, n) = (0, 1)
    while True: 
        (h, n) = (h + 1 / n, n + 1)
        yield h
h = harmonic() 
print(h) # generator object
print(next(h), next(h))
```

```python
# Handling Exceptions 
def divide_numbers(a,b): 
    result = a / b
    return result

# #ZeroDivisionError; Python crashes immediatley 
# print(divide_numbers(5,0))
# print("Done")

def divide_numbers(a,b): 
    try: 
        result = a / b 
        return result 
    except ZeroDivisionError: #don't have to specify the error 
        return "Error: Division by zero is not allowed."

print(divide_numbers(5,0))
print("Done")

#catch different errors 
def divide_numbers(a, b): 
    try: 
        result = a / b
        return result 
    except ZeroDivisionError: 
        return "Error: Division by zero is not allowed."
    except TypeError: 
        return "Error: Both arguments must be numbers."
    except: 
        return "An unexpected error occurred."

print(divide_numbers(3, 'a string'))

#can associate other blocsk with try block 
def deivide_numbers(a, b): 
    try: 
        result = a / b
    except ZeroDivisionError: 
        return "Error: Division by zero is not allowed."
    else: 
        return f"The result is {result}" 
    finally: 
        print("Execution completed.")

print(deivide_numbers(10, 2))
```
